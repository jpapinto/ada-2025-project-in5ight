```python
import matplotlib.pyplot as plt
ch = pew


# quick sanity
print(f"rows: {len(ch):,}, weeks: {ch['week_start'].nunique():,}")

# plot: 28d rolling efficiency (strict) + forward-filled (for smoother line)
plt.figure(figsize=(12,4))
plt.plot(ch["week_start"], ch["efficiency_4week"], label="efficiency_4week (28d rolling)")
plt.plot(ch["week_start"], ch["efficiency_4week_filled"], label="efficiency_28d_ffill4", linewidth=1, alpha=0.7)
plt.title("PewDiePie — Engagement Efficiency Over Time (28-day rolling)")
plt.xlabel("Week")
plt.ylabel("Efficiency (reactions / views)")
plt.legend()
plt.tight_layout()
plt.show()

# optional: context with activity (videos per week) on a secondary axis
fig, ax1 = plt.subplots(figsize=(12,3.6))
ax1.plot(ch["week_start"], ch["efficiency_28d_ffill4"], label="efficiency_28d_ffill4")
ax1.set_ylabel("Efficiency")
ax1.set_xlabel("Week")
ax2 = ax1.twinx()
ax2.bar(ch["week_start"], ch["videos_week"], alpha=0.25)
ax2.set_ylabel("Videos / week")
plt.title("Efficiency vs Activity")
fig.tight_layout()
plt.show()
```
